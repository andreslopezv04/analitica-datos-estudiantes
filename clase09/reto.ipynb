## Cómo se verifica un reto que no tiene respuesta correcta

Esto es distinto a los retos de las clases 2 a 5 y conviene ser exacto, porque de lo contrario
parecería que el cuaderno te está calificando la historia. No lo está haciendo.

En los retos anteriores había **una** respuesta correcta y el verificador comparaba una huella
digital de tu resultado contra la suya. Aquí no se puede, por dos razones: el dataset es tuyo (nadie
más sabe cuánto debe dar), y el entregable es una narrativa (no existe "el título correcto" de una
slide de implicación).

Así que la verificación está partida en tres, y cada parte dice en voz alta hasta dónde llega:

| Qué se comprueba | Cómo | Qué NO garantiza |
|------------------|------|------------------|
| **Las cifras existen y son usables** — `comprobar_cifra` | Que produjiste un número finito y legible, no un `None` ni una tabla | No sabe si el número es el correcto. Eso lo sabes tú |
| **La forma de los títulos** — `revisar_titulo`, `revisar_historia` | Que afirman algo, que caben en una línea, que el hallazgo lleva cifra, que la implicación no repite el hallazgo, que la acción concluye | No juzga si el título es **bueno**. Eso lo cierras tú en la autoevaluación del final |
| **La cifra de la slide se reproduce** — `comprobar_coherencia` | Que el número que escribiste en el título es el mismo que sale de recalcularlo desde el CSV | Nada que objetar: esta sí es automática y es la más importante del cuaderno |

**Por qué esto sigue siendo un reto y no un formulario.** Los verificadores mecánicos detectan las
fallas que se pueden detectar sin criterio: la plantilla sin llenar, el título que arranca como un
índice, la implicación que repite el hallazgo, la acción que no concluye. Son exactamente las cuatro
formas de fallar que más se repiten en el salón. Lo que **no** pueden decirte es si elegiste el
hallazgo que importa, si tu implicación le cambia algo a alguien, o si tu historia tiene tensión. Eso
es criterio, es lo que se evalúa en la sustentación del Momento 2, y es lo que estás entrenando hoy.

### El verificador

La celda de abajo define `comprobar_cifra(...)`, `revisar_titulo(...)`, `revisar_historia(...)` y
`comprobar_coherencia(...)`. Ejecútala una vez y sigue adelante: es andamiaje del curso, no materia

````python
# Verificador de las nueve tareas. Ejecute esta celda una vez y siga adelante.
# No hace falta entenderla hoy: es andamiaje del curso, no materia de la clase.
import hashlib
import unicodedata

_RESULTADOS = {}

_CLAVES = ["T1", "T2", "T3", "T4", "T5", "T6", "T7", "T8", "T9"]

_PISTAS = {
    "T1": "Tiene que ser UN numero, no una tabla ni un texto. Si le quedo un DataFrame, le falta quedarse con una celda: tabla.loc['fila', 'columna'], o una operacion que reduzca la tabla a un solo valor. Y si tiene mas de seis digitos, dividalo para que se pueda leer.",
    "T2": "Es un porcentaje, o sea una division por el total, por 100. Sale de la tabla que construyo con groupby: el valor de la fila que concentra el problema, dividido entre la suma de la columna completa.",
    "T3": "La segunda evidencia es otra cifra, distinta de T1, que sostenga la misma afirmacion desde otro angulo. Una razon (cuantas veces mas), una variacion entre dos periodos o una concentracion sirven. Si le sale NaN, revise si esta dividiendo entre cero o si la columna trae faltantes.",
    "T4": "Escriba el titulo entre comillas, como texto. De 4 a 18 palabras, que afirme algo. Si arranca con 'Analisis de', 'Resultados' o 'Introduccion', esta anunciando un tema en vez de instalar una pregunta.",
    "T5": "El titulo del hallazgo lleva la cifra ADENTRO. No 'encontramos diferencias importantes', sino la frase con el numero puesto. Es la unica de las cuatro donde la cifra es obligatoria.",
    "T6": "Si el verificador dice que comparte demasiadas palabras con el hallazgo, esta repitiendolo. Pregunte '¿y eso que?' sobre su propio hallazgo, tres veces seguidas, y escriba la respuesta de la tercera. Tiene que aparecer alguien a quien le cambia algo.",
    "T7": "La accion nombra QUE se hace y QUIEN lo hace. 'Se recomienda hacer seguimiento' y 'seguir investigando' son formas educadas de no concluir, y el verificador las conoce.",
    "T8": "Los cuatro titulos tienen que estar escritos y ser distintos entre si. Si falla por solapamiento, el problema esta en la slide 3: reescribala, no la slide 2.",
    "T9": "Las dos cifras tienen que ser el MISMO numero: la que escribio en el titulo del hallazgo y la que sale de recalcularla desde el CSV. Si no coinciden, averigue cual de las dos esta mal antes de presentar. Revise las unidades (¿dividio entre lo mismo?) y el filtro (¿esta contando las mismas filas?)."
}

_ESPERADO = {}


# --- Utilidades ------------------------------------------------------------

def _firma(valor):
    """Reduce un resultado a un texto reproducible, sin importar como se calculo."""
    if isinstance(valor, pd.DataFrame):
        partes = ["DataFrame", str(valor.shape), str([str(c) for c in valor.columns]),
                  str([str(i) for i in valor.index])]
        for columna in valor.columns:
            serie = valor[columna]
            if pd.api.types.is_bool_dtype(serie) or not pd.api.types.is_numeric_dtype(serie):
                partes.append(f"{columna}:{[str(v) for v in serie.tolist()]}")
            else:
                partes.append(f"{columna}:{round(float(serie.sum()), 4)}")
        return "|".join(partes)
    if isinstance(valor, pd.Series):
        return "|".join(["Series", str(len(valor)), str([str(i) for i in valor.index]),
                         str([str(v) for v in valor.tolist()])])
    if not isinstance(valor, str):
        try:
            return f"numero|{round(float(valor), 4)}"
        except (TypeError, ValueError):
            pass
    return f"otro|{valor!r}"


def _huella(valor):
    return hashlib.sha256(_firma(valor).encode("utf-8")).hexdigest()[:10]


def _redondear(valor, decimales):
    if decimales is None or valor is None:
        return valor
    if isinstance(valor, (pd.DataFrame, pd.Series)):
        return valor.round(decimales)
    try:
        return round(float(valor), decimales)
    except (TypeError, ValueError):
        return valor


def _normalizar(texto):
    """Minusculas, sin tildes y sin puntuacion. Para comparar sin falsos negativos."""
    plano = unicodedata.normalize("NFKD", str(texto).lower())
    plano = "".join(c for c in plano if not unicodedata.combining(c))
    return "".join(c if c.isalnum() or c.isspace() else " " for c in plano)


def _palabras_de_contenido(texto):
    """Las palabras que cargan significado: sin articulos, preposiciones ni numeros."""
    return {p for p in _normalizar(texto).split()
            if len(p) > 3 and p not in _PALABRAS_VACIAS}


# --- Comprobacion por huella (solo donde hay una respuesta correcta) --------

def comprobar(clave, valor, decimales=None):
    """Dice si el resultado es el correcto, sin revelar cual era."""
    _RESULTADOS[clave] = False
    if valor is None:
        print(f"[{clave}] Sin resolver todavia: la variable sigue valiendo None.")
        return
    valor = _redondear(valor, decimales)
    if isinstance(valor, pd.DataFrame):
        print(f"[{clave}] Usted produjo un DataFrame de {valor.shape[0]} filas "
              f"y {valor.shape[1]} columnas.")
    elif isinstance(valor, pd.Series):
        print(f"[{clave}] Usted produjo una Series de {len(valor)} elementos.")
    else:
        print(f"[{clave}] Usted produjo: {valor!r}")
    if _huella(valor) == _ESPERADO.get(clave):
        _RESULTADOS[clave] = True
        print(f"[{clave}] CORRECTO.")
    else:
        print(f"[{clave}] Todavia no coincide.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")


# --- Comprobacion por contrato (su dataset no lo conoce nadie mas) ----------

def comprobar_cifra(clave, valor, unidad=""):
    """Revisa que la cifra exista y sea utilizable. NO sabe cual es la correcta.

    Su dataset es suyo: no hay respuesta esperada contra la cual comparar. Lo que
    esto comprueba es que usted produjo un numero de verdad, finito y legible, y
    no un None, un DataFrame entero o un texto.
    """
    _RESULTADOS[clave] = False
    if valor is None:
        print(f"[{clave}] Sin resolver todavia: la variable sigue valiendo None.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
        return
    if isinstance(valor, (pd.DataFrame, pd.Series)):
        print(f"[{clave}] Eso es un {type(valor).__name__} completo, no una cifra. "
              f"Un titulo lleva un numero, no una tabla.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
        return
    try:
        numero = float(valor)
    except (TypeError, ValueError):
        print(f"[{clave}] Eso no es un numero: es {type(valor).__name__}.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
        return
    if numero != numero or numero in (float("inf"), float("-inf")):
        print(f"[{clave}] Salio NaN o infinito. Casi siempre es una division entre cero "
              f"o una columna con faltantes sin tratar.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
        return
    _RESULTADOS[clave] = True
    sufijo = f" {unidad}" if unidad else ""
    print(f"[{clave}] Registrada: {numero:,.2f}{sufijo}")
    if abs(numero) >= 1e6:
        print(f"[{clave}] AVISO: tiene {len(str(int(abs(numero))))} digitos. Nadie lee eso en una "
              f"slide. Divida entre 1e3 (miles), 1e6 (millones) o 1e12 (billones).")
    print(f"[{clave}] Este chequeo NO sabe si la cifra es la correcta: su dataset es de su "
          f"equipo. Lo unico que confirma es que la cifra existe y se puede leer.")


def comprobar_coherencia(clave, valor_recalculado, valor_declarado, tolerancia_relativa=0.01):
    """Compara la cifra que usted escribio en la slide contra la que produce el codigo.

    Esta si es una comprobacion automatica de verdad, y funciona sobre cualquier
    dataset: no necesita saber cual es la respuesta, solo que las dos coincidan.
    Es la regla de la clase convertida en chequeo: una cifra que no puede
    recalcular no va en una slide.
    """
    _RESULTADOS[clave] = False
    if valor_recalculado is None or valor_declarado is None:
        print(f"[{clave}] Sin resolver todavia: falta una de las dos cifras.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
        return
    try:
        a = float(valor_recalculado)
        b = float(valor_declarado)
    except (TypeError, ValueError):
        print(f"[{clave}] Las dos tienen que ser numeros.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
        return
    escala = max(abs(a), abs(b))
    diferencia = 0.0 if escala == 0 else abs(a - b) / escala
    print(f"[{clave}] Recalculada desde el CSV: {a:,.4f}")
    print(f"[{clave}] Declarada en su titulo:   {b:,.4f}")
    if diferencia <= tolerancia_relativa:
        _RESULTADOS[clave] = True
        print(f"[{clave}] CORRECTO: la cifra de su slide se reproduce desde los datos.")
    else:
        print(f"[{clave}] NO COINCIDEN: se apartan un {diferencia:.1%}.")
        print(f"[{clave}] Una de las dos esta mal, y hay que averiguar cual ANTES de presentar. "
              f"Si la que esta mal es la de la slide, corrijala; si es la del codigo, "
              f"revise el filtro o las unidades.")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")


# --- Revision mecanica de la narrativa -------------------------------------

_ETIQUETAS_VETADAS = (
    "analisis de", "resultados", "introduccion", "metodologia", "conclusiones",
    "estudio de", "datos de", "distribucion de", "comportamiento de",
    "estadisticas de", "informe de", "presentacion", "agenda", "objetivos",
    "marco teorico", "generalidades",
)

_FRASES_SIN_ACCION = (
    "seguir investigando", "hacer seguimiento", "profundizar en el analisis",
    "continuar el estudio", "tomar medidas", "mejorar el proceso",
    "se sugiere analizar", "futuros trabajos",
)

_MARCADORES_DE_PLANTILLA = (
    "tu titulo aqui", "escribe aqui", "escriba aqui", "version debil", "...",
)

_PALABRAS_VACIAS = {
    "a", "al", "algo", "ante", "asi", "aunque", "cada", "como", "con", "contra",
    "cual", "cuando", "de", "del", "desde", "donde", "dos", "el", "ella", "ellos", "en",
    "entre", "era", "es", "esa", "ese", "eso", "esta", "este", "esto", "estos", "ha",
    "hasta", "hay", "la", "las", "le", "les", "lo", "los", "mas", "me", "mi",
    "mientras", "muy", "ni", "no", "nos", "o", "para", "pero", "por", "porque", "que",
    "se", "segun", "ser", "si", "sin", "sobre", "solo", "son", "su", "sus", "tan",
    "te", "tiene", "todo", "todos", "un", "una", "uno", "unos", "y", "ya",
}

_AVISO_MECANICO = ("Esto es una revision MECANICA: mira la forma del titulo, no si el titulo "
                   "es bueno. Eso lo decide usted con los criterios de la autoevaluacion.")


def _faltas_de_forma(texto, con_cifra=False):
    faltas = []
    if texto is None or not str(texto).strip():
        return ["esta vacio: escriba el titulo entre comillas"]
    limpio = str(texto).strip()
    plano = _normalizar(limpio).strip()
    if any(marca in plano or marca in limpio for marca in _MARCADORES_DE_PLANTILLA):
        faltas.append("sigue siendo el texto de la plantilla, no un titulo suyo")
    palabras = plano.split()
    if len(palabras) < 4:
        faltas.append(f"tiene {len(palabras)} palabras: un titulo de menos de 4 es una etiqueta")
    if len(palabras) > 18:
        faltas.append(f"tiene {len(palabras)} palabras: si no cabe en una linea, no es un titulo")
    for etiqueta in _ETIQUETAS_VETADAS:
        if plano.startswith(etiqueta):
            faltas.append(f"arranca como un indice ('{etiqueta}...'): eso anuncia un tema, "
                          f"no afirma nada")
            break
    if limpio.endswith(":"):
        faltas.append("termina en dos puntos: eso es el encabezado de una lista, no una afirmacion")
    if con_cifra and not any(c.isdigit() for c in limpio):
        faltas.append("no tiene ninguna cifra: el titulo del hallazgo lleva el numero adentro")
    return faltas


def revisar_titulo(clave, texto, con_cifra=False):
    """Revisa la FORMA de un titulo de slide. No juzga el contenido."""
    _RESULTADOS[clave] = False
    faltas = _faltas_de_forma(texto, con_cifra=con_cifra)
    if texto:
        print(f"[{clave}] Su titulo: \"{str(texto).strip()}\"")
    if faltas:
        print(f"[{clave}] Todavia no pasa la revision de forma:")
        for falta in faltas:
            print(f"[{clave}]   - {falta}")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
    else:
        _RESULTADOS[clave] = True
        print(f"[{clave}] La forma esta bien: afirma algo, cabe en una linea y no es una etiqueta.")
    print(f"[{clave}] {_AVISO_MECANICO}")


def revisar_historia(clave, contexto, hallazgo, implicacion, accion):
    """Revisa los cuatro titulos juntos: las fallas que solo se ven al leerlos seguidos."""
    _RESULTADOS[clave] = False
    titulos = {"contexto": contexto, "hallazgo": hallazgo,
               "implicacion": implicacion, "accion": accion}
    faltas = [f"falta el titulo de la slide de {n}" for n, t in titulos.items()
              if t is None or not str(t).strip()]

    if not faltas:
        print(f"[{clave}] Sus cuatro titulos, seguidos y sin el cuerpo:")
        for i, (nombre, texto) in enumerate(titulos.items(), start=1):
            print(f"[{clave}]   {i}. {str(texto).strip()}")

        planos = {n: _normalizar(t).strip() for n, t in titulos.items()}
        vistos = {}
        for nombre, plano in planos.items():
            if plano in vistos:
                faltas.append(f"los titulos de {vistos[plano]} y {nombre} son el mismo texto")
            vistos[plano] = nombre

        palabras_hallazgo = _palabras_de_contenido(hallazgo)
        palabras_implicacion = _palabras_de_contenido(implicacion)
        union = palabras_hallazgo | palabras_implicacion
        if union:
            solape = len(palabras_hallazgo & palabras_implicacion) / len(union)
            if solape >= 0.5:
                faltas.append(
                    f"la implicacion comparte el {solape:.0%} de sus palabras con el hallazgo: "
                    f"probablemente lo esta repitiendo con otras palabras. Preguntese "
                    f"'¿y eso que?' hasta que la frase deje de ser un dato y sea una "
                    f"consecuencia para alguien")

        for frase in _FRASES_SIN_ACCION:
            if frase in planos["accion"]:
                faltas.append(f"la accion dice '{frase}', que es una forma educada de no "
                              f"concluir: nombre que se hace y quien lo hace")
                break

    if faltas:
        print(f"[{clave}] Todavia no pasa:")
        for falta in faltas:
            print(f"[{clave}]   - {falta}")
        print(f"[{clave}] Pista: {_PISTAS[clave]}")
    else:
        _RESULTADOS[clave] = True
        print(f"[{clave}] Los cuatro titulos pasan la revision mecanica.")
    print(f"[{clave}] {_AVISO_MECANICO} Que esa secuencia SUENE a historia y no a indice "
          f"lo decide usted, leyendola en voz alta.")


def resumen_puntos_de_control():
    """Estado de los puntos de control del cuaderno."""
    print("Punto de control")
    print("-" * 46)
    for clave in _CLAVES:
        estado = "correcto" if _RESULTADOS.get(clave) else "pendiente"
        print(f"  {clave}: {estado}")
    logrados = sum(1 for c in _CLAVES if _RESULTADOS.get(c))
    print("-" * 46)
    print(f"{logrados} de {len(_CLAVES)} en verde.")
    print("Los puntos narrativos son mecanicos: verde significa 'tiene forma de titulo',")
    print("no significa 'es un buen titulo'. Esa parte la cierra la autoevaluacion final.")


print("Verificador listo. Ninguna tarea de hoy se comprueba por huella: lea la tabla de arriba.")
````

# Formulario de feedback: I like, I wish, What if

Diligéncialo para cada equipo que escuches. Tres frases, en este orden y sin saltarse ninguna.

- **I like...** qué funcionó. Concreto, no "estuvo bien".
- **I wish...** qué faltó. Sobre el trabajo, nunca sobre la persona.
- **What if...** una propuesta, no una queja.

Ejemplo de los tres, sobre un mismo pitch:

- "I like que arrancaste con la cifra de la brecha: entendí el problema en 10 segundos."
- "I wish el título de la tercera slide dijera qué concluir; me tocó deducirlo."
- "What if muestras el gráfico por municipio antes del total? Creo que el contraste pega más fuerte."

**El que presenta no defiende.** Escucha y anota. Discutir el feedback se come el tiempo del
siguiente equipo.
