## Paso 3 de 5 — Leer un CSV real del repositorio

Se lee `datos/HISTORICO_CONSUMO.csv`: el consumo de agua de Empocaldas, el dataset que acompaña
todo el semestre.

La ruta es **relativa**: se cuenta desde la carpeta donde está este notebook, que es la raíz del
repositorio. Si sale `FileNotFoundError`, casi siempre es que abrió el notebook desde otra carpeta.

```python
df = pd.read_csv("datos/HISTORICO_CONSUMO.csv")

print("Filas:   ", df.shape[0])
print("Columnas:", df.shape[1])
print("\nNombres de las columnas:")
for columna in df.columns:
    print(" -", columna)
```

Un `DataFrame` es la tabla con la que se trabaja todo el semestre: filas y columnas, como una hoja

## Paso 4 de 5 — Pintar un gráfico

Si aparece un gráfico de barras debajo de esta celda, matplotlib y seaborn quedaron bien instalados
y el notebook puede mostrar imágenes.

```python
conteo = df["MUNICIPIO"].value_counts().head(10)

plt.figure(figsize=(9, 4))
sns.barplot(x=conteo.values, y=conteo.index, color="#2f6f9f")
plt.title("Registros por municipio (los 10 más frecuentes)")
plt.xlabel("Número de registros")
plt.ylabel("")
plt.tight_layout()
plt.show()
```
